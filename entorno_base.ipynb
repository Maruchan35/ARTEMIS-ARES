# Entorno Base — Colab + Google Drive

Flujo de trabajo recomendado:
1. Usa **CPU** mientras preparamos y editamos codigo.
2. Cambia a **GPU T4** para pruebas con CUDA o **A100** para entrenamiento/inferencia pesada.
3. Todo el proyecto vive en `/content/drive/MyDrive/colab_entorno`, asi que **no se borra** al cambiar o reiniciar el entorno.

```python
from google.colab import drive
import os
import sys

drive.mount('/content/drive')

PROJECT_DIR = '/content/drive/MyDrive/colab_entorno'
SRC_DIR = os.path.join(PROJECT_DIR, 'src')
DATA_DIR = os.path.join(PROJECT_DIR, 'data')
OUTPUT_DIR = os.path.join(PROJECT_DIR, 'outputs')

for folder in [SRC_DIR, DATA_DIR, OUTPUT_DIR]:
    os.makedirs(folder, exist_ok=True)

os.chdir(PROJECT_DIR)
if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

print(f'Directorio activo: {os.getcwd()}')
```

```python
import shutil
import subprocess
import psutil

ram_gb = psutil.virtual_memory().total / (1024**3)
disk_free_gb = shutil.disk_usage('/content').free / (1024**3)
print(f'RAM total: {ram_gb:.1f} GB | Disco libre (/content): {disk_free_gb:.1f} GB')

res = subprocess.run(['nvidia-smi'], capture_output=True, text=True)
if res.returncode == 0:
    print(res.stdout)
else:
    print('Sin GPU activa (modo CPU). Ideal para preparar codigo con minimo consumo.')
```
